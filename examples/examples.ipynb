```python
filename = 'example.cbf'
file_descriptor = open(filename, 'rb')
content = file_descriptor.read()
index = content.find(b'\x0C\x1A\x04\xD5')
file_descriptor.close()
```

```python
print(bytes.decode(content[:index]))
```

```python
# print cbf header
for line in content[:index].splitlines():
    print(bytes.decode(line))
```

```python
import re
file_header = content[:index]
pattern = re.compile(r'X-Binary-([\w-]*)([\s:]*)(.*)')
header = dict()
for line in file_header.splitlines():
    m = pattern.search(bytes.decode(line))
    if m:
        key = m.group(1).lower().replace('-', '_')
        val = m.group(3).strip()
        try:
            val = int(val)
        except ValueError:
            try:
                val = float(val)
            except ValueError:
                val = val
        print(key, val)
        header[key] = val
print(header)
```
